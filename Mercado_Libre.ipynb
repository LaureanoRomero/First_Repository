```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import os

class Scraper():

    def menu(self):
        menu = ("""
    Escoge el país:
    1. Argentina
    2. Bolivia
    3. Brasil
    4. Chile
    5. Colombia
    6. Costa Rica
    7. Dominicana
    8. Ecuador
    9. Guatemala
    10. Honduras
    11. México
    12. Nicaragua
    13. Panamá
    14. Paraguay
    15. Perú
    16. Salvador
    17. Uruguay
    18. Venezuela
        """)

        valid_options = list(range(1, 19))

        while True:
            print(menu)
            opcion = int(input('Número de país (Ejemplo: 5): '))

            if opcion in valid_options:

                urls = {
                1: 'https://listado.mercadolibre.com.ar/',
                2: 'https://listado.mercadolibre.com.bo/',
                3: 'https://listado.mercadolibre.com.br/',
                4: 'https://listado.mercadolibre.cl/',
                5: 'https://listado.mercadolibre.com.co/',
                6: 'https://listado.mercadolibre.com.cr/',
                7: 'https://listado.mercadolibre.com.do/',
                8: 'https://listado.mercadolibre.com.ec/',
                9: 'https://listado.mercadolibre.com.gt/',
                10: 'https://listado.mercadolibre.com.hn/',
                11: 'https://listado.mercadolibre.com.mx/',
                12: 'https://listado.mercadolibre.com.ni/',
                13: 'https://listado.mercadolibre.com.pa/',
                14: 'https://listado.mercadolibre.com.py/',
                15: 'https://listado.mercadolibre.com.pe/',
                16: 'https://listado.mercadolibre.com.sv/',
                17: 'https://listado.mercadolibre.com.uy/',
                18: 'https://listado.mercadolibre.com.ve/',
                }

                self.base_url = urls[opcion]
                break
            else:
                print("Escoge un número del 1 al 18")



    def scraping(self):
        # User search
        product_name = input("\nProducto: ")
        # Clean the user input
        cleaned_name = product_name.replace(" ", "-").lower()
        # Create the urls to scrap
        urls = [self.base_url + cleaned_name]

        page_number = 50
        for i in range(0, 10000, 50):
            urls.append(f"{self.base_url}{cleaned_name}_Desde_{page_number + 1}_NoIndex_True")
            page_number += 50

        # create a list to save the data
        self.data = []
        # create counter
        c = 1

        # Iterate over each url
        for i, url in enumerate(urls, start=1):

            # Get the html of the page
            response = requests.get(url)
            soup = BeautifulSoup(response.text, 'html.parser')

            # take all posts
            content = soup.find_all('li', class_='ui-search-layout__item')

            # Check if there's no content to scrape
            if not content:
                print("\nTermino el scraping.")
                break

            print(f"\nScrapeando pagina numero {i}. {url}")


            # iteration to scrape posts
            for post in content:
                # get the title
                title = post.find('h2').text
                # get the price
                price = post.find('span', class_='andes-money-amount__fraction').text
                # Precio descuento <span class="ui-search-price__discount">25% OFF</span>
                perc_desc = post.find('span', class_='ui-search-price__discount')
                # Calificaciones y nota
                count = post.find('span', class_="andes-visually-hidden")
                #count =  <span class="ui-search-reviews__amount" aria-hidden="true">(71)</span>

                # get the url post
                post_link = post.find("a")["href"]
                # get the url image
                try:
                    img_link = post.find("img")["data-src"]
                except:
                    img_link = post.find("img")["src"]

                # show the data already scraped
                # print(f"{c}. {title}, {price}, {post_link}, {img_link}")

                # save in a dictionary
                post_data = {
                    "title": title,
                    "price": price,
                    "perc_desc":perc_desc,
                    "count": count,
                    "post link": post_link,
                    "image link": img_link
                }
                # save the dictionaries in a list
                self.data.append(post_data)
                c += 1

    def export_to_csv(self):
        # export to a csv file
        df = pd.DataFrame(self.data)
        df.to_csv(r"/content/sample_data/mercadolibre_scraped_data.csv", sep=";")

if __name__ == "__main__":
    s = Scraper()
    s.menu()
    s.scraping()
    s.export_to_csv()

# <span class="ui-pdp-subtitle">Nuevo  |  +5 vendidos</span>
# <span class="andes-visually-hidden">Calificación 4.7 de 5. 71 opiniones.</span>
```

```python
# Lee el archivo CSV y crea un DataFrame
df = pd.read_csv('/content/sample_data/mercadolibre_scraped_data.csv', delimiter=";")
df
```

```python
import re

# Función para extraer las calificaciones y opiniones usando expresiones regulares
def extraer_calificacion_opiniones(texto):
    if isinstance(texto, str):  # Verifica si es una cadena de texto
        calificacion_match = re.search(r'Calificación (\d+\.\d+)', texto)
        opiniones_match = re.search(r'(\d+)\s*opiniones?', texto)

        calificacion = calificacion_match.group(1) if calificacion_match else None
        opiniones = int(opiniones_match.group(1)) if opiniones_match else None

        return calificacion, opiniones
    else:
        return None, None

# Aplicar la función a la columna 'count' y crear nuevas columnas 'calificacion' y 'opiniones'
df[['calificacion', 'opiniones']] = df['count'].apply(lambda x: pd.Series(extraer_calificacion_opiniones(x)))

# Imprimir el DataFrame resultante
print(df)
```

```python
import numpy as np
# Función para extraer el descuento de la cadena usando expresiones regulares
def extraer_descuento(cadena):
    if pd.isnull(cadena) or not isinstance(cadena, str):
        return np.nan
    else:
        # Busca el patrón de números seguido por '% OFF' y extrae el número
        match = re.search(r'(\d+)% OFF', cadena)
        if match:
            return int(match.group(1))  # Devuelve el número encontrado
        else:
            return np.nan

# Aplicar la función a la columna 'count' y crear una nueva columna 'descuento'
df['descuento'] = df['perc_desc'].apply(extraer_descuento)

# Mostrar el DataFrame con la nueva columna 'descuento'
print(df)
```

```python
# Convertir las columnas 'price' y 'calificacion' a tipo float64
#df['price'] = df['price'].astype('float64')
df['price'] = df['price'].str.replace('.', '').astype('float64')
df['calificacion'] = df['calificacion'].astype('float64')
```
